## Introduction

This project analyses 14 days of sales and the current
stock position of Wanjiku's fresh-produce stall at
Gikomba Market. The objective is to evaluate business
performance, identify sales trends, classify daily
performance, assess stock requirements, and provide
actionable business recommendations.

```python
#daily sales for 14 days
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,2500, 2200, 3100, 2400, 3900, 5600, 4700] # 14 days, KES

# Days of the week repeated twice
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2
#qty = kg in stock, reorder level (kg), cost/price = KES per kg

#stock information
#qty = quantity in kg
#reorder = reorder level in kg
#cost = buying price per kg
#price = selling price per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":{"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
```

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

print(f"Total Sales: {total_sales}")
print(f"Average Daily Sales: {average_sales:,.2f}")
```

```python
#find highest and lowest sales
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

#find their positions in the list
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

#find the week number
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

#Dsiplay results
print(f"Highest Sales: {highest_sales}")
print(f"Best day: {days[best_index]}")
print(f"Best week: {best_week}")

print(f"\n Lowest Sales: {lowest_sales}")
print(f"Worst Day: {days[worst_index]}")
print(f"Worst Week: {worst_week}")
```

```python
week1 = daily_sales[:7]
week2 = daily_sales[7:]
```

```python
def week_totals(sales):
  """Calculate total sales for Week 1 and week 2.
  Returns:
  A tuple containing week 1 and 2 sales totals """

  week1_total = sum(sales[:7])
  week2_total = sum(sales[7:])

  return week1_total, week2_total

#Get weekly sales total
week1, week2 = week_totals(daily_sales)

#Calculate week-on-week growth percentage
growth = ((week2 - week1) / week1) * 100

#Display results
print(f"Week 1 Sales: KES {week1:,.2f}")
print(f"Week 2 Sales: KES {week2:,.2f}")
print(f"Week-on-week Growth: {growth:,.2f}%")
```
