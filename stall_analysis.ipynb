## Introduction

This project analyses 14 days of sales and the current
stock position of Wanjiku's fresh-produce stall at
Gikomba Market. The objective is to evaluate business
performance, identify sales trends, classify daily
performance, assess stock requirements, and provide
actionable business recommendations.

```python
#daily sales for 14 days
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,2500, 2200, 3100, 2400, 3900, 5600, 4700] # 14 days, KES

# Days of the week repeated twice
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2
#qty = kg in stock, reorder level (kg), cost/price = KES per kg

#stock information
#qty = quantity in kg
#reorder = reorder level in kg
#cost = buying price per kg
#price = selling price per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":{"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
```

```python

#Task 1 Sales Performance
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

print(f"Total Sales: {total_sales}")
print(f"Average Daily Sales: {average_sales:,.2f}")
```

```python
#find highest and lowest sales
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

#find their positions in the list
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

#find the week number
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

#Dsiplay results
print(f"Highest Sales: {highest_sales}")
print(f"Best day: {days[best_index]}")
print(f"Best week: {best_week}")

print(f"\n Lowest Sales: {lowest_sales}")
print(f"Worst Day: {days[worst_index]}")
print(f"Worst Week: {worst_week}")
```

```python
week1 = daily_sales[:7]
week2 = daily_sales[7:]
```

```python
#Weekly Sales Analysys
def week_totals(sales):
  """Calculate total sales for Week 1 and week 2.
  Returns:
  A tuple containing week 1 and 2 sales totals """

  week1_total = sum(sales[:7])
  week2_total = sum(sales[7:])

  return week1_total, week2_total

#Get weekly sales total
week1, week2 = week_totals(daily_sales)

#Calculate week-on-week growth percentage
growth = ((week2 - week1) / week1) * 100

#Display results
print(f"Week 1 Sales: KES {week1:,.2f}")
print(f"Week 2 Sales: KES {week2:,.2f}")
print(f"Week-on-week Growth: {growth:,.2f}%")
```

```python
stronghold_threshold = average_sales * 1.20
weak_threshold = average_sales * 0.80
```

```python
def classify_day(amount, average):
  """Classify daily sales as strong, Normal or Weak
      Strong: More than 20% above average
      Weak: More than 20% below average
      Normal: Within the 20% boundaries"""

  if amount > average:
    return "Strong"

  elif amount < average * 0.80:
    return "Weak"
  else:
    return "Normal"
```

```python
category_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

#classify each day's sales
daily_categories = []
for sale in daily_sales:
  category = classify_day(sale, average_sales)
  daily_categories.append(category)
  category_counts [category] += 1


#Display classifications
for day, sale, category in zip(days, daily_sales, daily_categories):
  print(f"{day}: KES {sale:,} - {category}")

#Display category totals
print("\nCategory Summary")

for category, count in category_counts.items():
  print(f"{category}: {count} days")
```

```python
#Task 4
#Weekday vs Weekend Analysis
# Collect weekday and weekend sales
weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):

    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])

    else:
        weekend_sales.append(daily_sales[i])

# Calculate average sales
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

# Display results
print(f"Weekday Average: KES {weekday_average:,.2f}")
print(f"Weekend Average: KES {weekend_average:,.2f}")
print(f"Percentage Difference: {percentage_difference:.2f}%")
```

```python
def get_reorder_products(stock):
  """Return product whose cureent quanity is at or below their reorder level"""
  reorder_products = []
  for product, details in stock.items():
    if details["qty"] <= details["reorder"]:
      reorder_products.append(product)

  return reorder_products
```

```python
#Task 5
#Inventory and Gross Margin Analysis
# Calculate gross margin for each product

gross_margins = {}

for product, details in stock.items():

    cost = details["cost"]
    price = details["price"]

    margin = ((price - cost) / price) * 100

    gross_margins[product] = margin

# Display gross margins
print("GROSS MARGINS")

for product, margin in gross_margins.items():
    print(f"{product}: {margin:.2f}%")
```

```python
products_to_reorder = get_reorder_products(stock)
print("PRODUCT REQUIRING RESTOCKING")

for product in products_to_reorder:
  print(
      f"{product}: Current stock = {stock[product]['qty']} kg, "
      f"Reorder level = {stock[product]['reorder']} kg"
  )
```
